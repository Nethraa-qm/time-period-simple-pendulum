```python
import math
def time_period_of_pendulum (length, celestial_body):
    if celestial_body == "Earth" :
        g = 9.8 
    elif celestial_body == "Mars" :
        g = 3.72
    elif celestial_body == "Moon" :
        g = 1.625
    elif celestial_body == "Jupiter" :
        g = 24.79
    elif celestial_body == "Saturn" :
        g = 10.44
    elif celestial_body == "Uranus" :
        g = 8.69
    elif celestial_body == "Neptune" :
        g = 11.15
    elif celestial_body == "Mercury" :
        g = 3.7
    elif celestial_body == "Venus" :
        g = 8.87
    elif celestial_body == "Pluto" :
        g = 0.62
    elif celestial_body == "Sun" :
        g = 274
    else :
        print ("Sorry, data for given celestial_body is not available")
        return
    time_period = 2*(math.pi)*(math.sqrt(length / g))
    
    print (f"Time period on {celestial_body} is {time_period:.2f}s")
  
time_period_of_pendulum(1,"Venus")
    
```
